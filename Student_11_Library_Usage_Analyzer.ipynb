## 6. Your individual project challenge
**Challenge:** Calculate return rate; identify students with most late returns; summarise usage.

Write your solution below and explain what the result means.


```python
# TODO: Solve your assigned challenge
# Calculate return rate for each student
df["Return_Rate"] = np.where(
    df["Books_Issued"] > 0,
    (df["Books_Returned"] / df["Books_Issued"]) * 100,
    0
)

df["Return_Rate"] = df["Return_Rate"].round(2)
df["Pending_Books"] = (
    df["Books_Issued"] - df["Books_Returned"]
)

print("=== STUDENT RETURN RATE ===")
display(df[["Student_ID", "Name", "Return_Rate"]])

max_late = df["Late_Returns"].max()

most_late_students = df[
    df["Late_Returns"] == max_late
]

print("\n=== STUDENTS WITH MOST LATE RETURNS ===")
display(
    most_late_students[
        ["Student_ID", "Name", "Late_Returns"]
    ]
)
print("\n=== LIBRARY USAGE SUMMARY ===")

print("Total Students:", len(df))

print("Total Books Issued:",
      df["Books_Issued"].sum())

print("Total Books Returned:",
      df["Books_Returned"].sum())

print("Total Pending Books:",
      df["Pending_Books"].sum())

print("Total Late Returns:",
      df["Late_Returns"].sum())

total_issued = df["Books_Issued"].sum()
total_returned = df["Books_Returned"].sum()

overall_return_rate = (
    total_returned / total_issued * 100
    if total_issued > 0 else 0
)

print("Overall Return Rate:",
      round(overall_return_rate, 2), "%")
```
